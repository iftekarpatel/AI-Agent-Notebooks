# Import Libraries

```python
import os
import google.generativeai as genai
from dotenv import load_dotenv
from IPython.display import Markdown, display
```

# Load Gemini API Key

```python
load_dotenv()

gemini_api_key = os.getenv("GEMINI_API_KEY")

print("First 5 characters:", gemini_api_key[:5])

genai.configure(api_key=gemini_api_key)
```

# Create Helper Markdown Function

```python
def print_markdown(text):
    display(Markdown(text))
```

# Create Gemini Model Object

```python
model = genai.GenerativeModel("gemini-2.5-flash-lite")
```

# Define Fact Checker Instructions

```python
fact_checker_prompt = """
You are a fact checker who verifies the accuracy of statements.

When given a statement:
- Carefully analyze its factual accuracy.
- Respond with:
    - A verdict prefix: ✅ True or ❌ False
    - A brief one-sentence explanation.
"""
```

# Create Statement

```python
statement = "The Great Wall of China is visible from space with the naked eye."
```

# Send Prompt to Gemini

```python
full_prompt = fact_checker_prompt + "\n\nStatement: " + statement

response = model.generate_content(full_prompt)

print_markdown("### 🤖 Agent's Response:\n")
print_markdown(response.text)
```
